```python
# Inicializamos una lista vacía para almacenar nuestros contactos.
# Cada contacto será un diccionario con claves como 'nombre', 'telefono', 'email'.
contactos = [] # <--- CORRECCIÓN: Usar 'contactos' consistentemente

# --- Bucle principal del programa ---
while True:
    print("\n--- Gestor de Contactos ---") # <--- CORRECCIÓN: Nombre del menú
    print("1. Agregar contacto")
    print("2. Ver contactos")
    print("3. Buscar contacto")
    print("4. Actualizar contacto")
    print("5. Eliminar contacto")
    print("6. Salir")

    opcion = input("Elige una opción: ")

    if opcion == '1':
        print("\n--- Agregar Contacto ---")
        nombre = input("Nombre: ").strip().title()
        telefono = input("Teléfono: ").strip()
        email = input("Email: ").strip()

        # Validar que al menos el nombre no esté vacío
        if not nombre:
            print("El nombre no puede estar vacío. Contacto no agregado.") # <--- CORRECCIÓN: Guardado a agregado
            continue # Vuelve al inicio del bucle.

        # Crear un diccionario para el nuevo contacto.
        nuevo_contacto = {
            "nombre": nombre,
            "telefono": telefono,
            "email": email
        }
        contactos.append(nuevo_contacto) # <--- CORRECCIÓN: 'contacto' a 'contactos'
        print(f"Contacto '{nombre}' agregado exitosamente.") # <--- CORRECCIÓN: 'ecitosamente' a 'exitosamente'

    elif opcion == '2':
        print("\n--- Ver Contactos ---")
        if not contactos: # <--- CORRECCIÓN: 'contacto' a 'contactos'
            print("No hay contactos en la agenda.")
        else:
            for i, contacto_actual in enumerate(contactos, 1): # <--- CORRECCIÓN: 'contactos' a 'contacto_actual' para claridad
                # Accedemos a los valores del diccionario usando sus claves
                print(f"{i}. Nombre: {contacto_actual['nombre']}, Teléfono: {contacto_actual['telefono']}, Email: {contacto_actual['email']}")

    elif opcion == '3':
        print("\n--- Buscar Contacto ---") # <--- CORRECCIÓN: 'Contactos' a 'Contacto'
        if not contactos: # <--- CORRECCIÓN: 'contacto' a 'contactos'
            print("No hay contactos para buscar.")
            continue

        nombre_buscar = input("Ingresa el nombre (o parte del nombre) a buscar: ").strip().title() # <--- CORRECCIÓN: 'Ingresar' a 'Ingresa'
        encontrados = [] # Lista para almacenar los contactos que coincidan

        for contacto_actual in contactos: # <--- CORRECCIÓN: 'contactos' a 'contacto_actual'
            # Usamos 'in' para ver si la parte del nombre a buscar está en el nombre del contacto
            if nombre_buscar in contacto_actual['nombre']:
                encontrados.append(contacto_actual)

        if encontrados:
            print("\n--- Contactos Encontrados ---")
            for i, contacto_actual in enumerate(encontrados, 1): # <--- CORRECCIÓN: 'contactos' a 'contacto_actual'
                print(f"{i}. Nombre: {contacto_actual['nombre']}, Teléfono: {contacto_actual['telefono']}, Email: {contacto_actual['email']}")
        else:
            print(f"No se encontraron contactos con '{nombre_buscar}'.") # <--- CORRECCIÓN: 'conntactos' a 'contactos'

    elif opcion == '4':
        print("\n--- Actualizar Contacto ---")
        if not contactos: # <--- CORRECCIÓN: 'contactos' a 'contactos'
            print("No hay contactos para actualizar.")
            continue

        nombre_actualizar = input("Ingresa el nombre del contacto a actualizar: ").strip().title() # <--- CORRECCIÓN: 'Ingrese' a 'Ingresa'
        contacto_encontrado = None # Para guardar el contacto si lo encontramos

        # Buscamos el contacto por nombre exacto
        for contacto_actual in contactos: # <--- CORRECCIÓN: 'contactos' a 'contacto_actual'
            if contacto_actual['nombre'] == nombre_actualizar: # <--- CORRECCIÓN: 'contacto' a 'contacto_actual'
                contacto_encontrado = contacto_actual
                break # Salimos del bucle una vez que lo encontramos

        if contacto_encontrado:
            print(f"Editando contacto: {contacto_encontrado['nombre']} (Tel: {contacto_encontrado['telefono']}, Email: {contacto_encontrado['email']})")
            print("Deja en blanco un campo si no quieres modificarlo.") # <--- CORRECCIÓN: 'el campo' a 'un campo'

            nuevo_telefono = input(f"Nuevo teléfono (actual: {contacto_encontrado['telefono']}): ").strip() # <--- CORRECCIÓN: paréntesis en 'actual'
            nuevo_email = input(f"Nuevo email (actual: {contacto_encontrado['email']}): ").strip()

            if nuevo_telefono:
                contacto_encontrado['telefono'] = nuevo_telefono
            if nuevo_email:
                contacto_encontrado['email'] = nuevo_email

            print(f"Contacto '{nombre_actualizar}' actualizado exitosamente.") # <--- INDENTACIÓN CORREGIDA
        else: # <--- INDENTACIÓN CORREGIDA
            print(f"Contacto '{nombre_actualizar}' no encontrado.")


    elif opcion == '5': # <--- INDENTACIÓN CORREGIDA
        print("\n--- Eliminar Contacto ---")
        if not contactos: # <--- CORRECCIÓN: 'contacto' a 'contactos'
            print("No hay contactos para eliminar.")
            continue # <--- Añadido continue para salir si no hay contactos

        nombre_eliminar = input("Ingresa el nombre del contacto a eliminar: ").strip().title()

        # Usamos una lista de comprensión para crear una nueva lista
        # que contenga solo los contactos que NO coincidan con el nombre a eliminar.
        # Esto es una forma segura y eficiente de eliminar elementos.
        contactos_original = contactos[:] # <--- CORRECCIÓN: 'contacto' a 'contactos'
        contactos = [contacto_actual for contacto_actual in contactos if contacto_actual['nombre'] != nombre_eliminar] # <--- CORRECCIÓN: 'contactos' a 'contacto_actual' y 'contacto' a 'contactos'

        if len(contactos) < len(contactos_original):
            print(f"Contacto '{nombre_eliminar}' eliminado exitosamente.")
        else:
            print(f"Contacto '{nombre_eliminar}' no encontrado.")

    elif opcion == '6': # <--- INDENTACIÓN CORREGIDA
        print("¡Gracias por usar el Gestor de Contactos! ¡Hasta luego!")
        break
    else: # <--- INDENTACIÓN CORREGIDA
        print("Opción no válida. Por favor, elige un número del 1 al 6.")
```
